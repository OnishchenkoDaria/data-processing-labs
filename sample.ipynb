# TASK 1 : PREPARING

```python
import numpy as np
import pandas as pd
```

## 1.1 -- Importing dataset

```python
df = pd.read_csv("./data/housewares.csv")
df
```

```python
df.head(5)
```

## 1.2 -- Filtering input

```python
df = df[['Name', 'Variation', 'Pattern', 'DIY', 'Pattern Customize', 'Kit Cost', 'Sell', 'Source', 'Filename', 'Internal ID']]
df
```

## 1.3 -- Removing empty data

```python
df.isna().sum()
```

```python
df = df[df['Variation'].notna()]
df = df[df['Pattern'].notna()]
df = df[df['Kit Cost'].notna()]
df
```

## 1.4 -- Formatting data

```python
df.dtypes
```

```python
df['Pattern Customize'] = df['Pattern Customize'].astype(bool)
df['DIY'] = df['DIY'].astype(bool)
df.dtypes
```

## 2.6 -- Comparing data groups

```python
pink_items = df[df['Variation'] == 'Pink']
print_items = df[df['Pattern'].str.contains('print', case=False, na=False)]
```

```python
summary_data = pd.DataFrame({
    'category_name': ['Pink colored items', 'Items with specific print'],
    'total_records': [len(pink_items), len(print_items)],
})
summary_data
```

```python
avg_pink_items = pink_items['Sell'].mean()
avg_print_items = print_items['Sell'].mean()
```

```python
summary_data['avg_price'] = [avg_pink_items, avg_print_items]
summary_data
```
